````python
def get_relevant_relationships(db, selected_columns):

    selected_set = set(selected_columns)
    relationships = []

    for fk in db["foreign_keys"]:

        col1 = db["column_names"][fk[0]]
        col2 = db["column_names"][fk[1]]

        table1 = db["table_names"][col1[0]]
        table2 = db["table_names"][col2[0]]

        full_col1 = f"{table1}.{col1[1]}"
        full_col2 = f"{table2}.{col2[1]}"

        if full_col1 in selected_set and full_col2 in selected_set:
            relationships.append(
                f"{full_col1} → {full_col2}"
            )

    return relationships
````

````python
def create_schema_context(db, selected_columns, relationships):

    schema_text = f"Database: {db['db_id']}\n\n"

    schema_text += "Relevant Tables and Columns:\n"

    for table_idx, table in enumerate(db["table_names_original"]):

        table_columns = []

        for col_idx, column in enumerate(db["column_names"]):

            if column[0] != table_idx or column[1] == "*":
                continue

            # Name used by retrieval
            normalized_name = f"{db['table_names'][table_idx]}.{column[1]}"

            # Actual database name used for SQL generation
            original_name = db["column_names_original"][col_idx][1]

            if normalized_name in selected_columns:
                table_columns.append(original_name)

        if table_columns:
            schema_text += f"\nTable: {table}\n"

            for column in table_columns:
                schema_text += f"- {column}\n"

    if relationships:
        schema_text += "\nRelationships:\n"

        for relationship in relationships:
            schema_text += f"- {relationship}\n"

    return schema_text
````

# Function for complete pipeline

````python
# Create the preprocessing function
def preprocess_question(question):
    # 1. Convert to lowercase
    question = question.lower()

    # 2. Tokenize the question
    tokens = nltk.word_tokenize(question)

    # 3. Identify parts of speech
    tagged_tokens = pos_tag(tokens)

    # 4. Lemmatize using POS tags
    normalized_words = [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tagged_tokens
        if word.isalpha()
    ]

    # 5. Remove stopwords
    filtered_words = [
        word for word in normalized_words
        if word not in stop_words
    ]

    return filtered_words
````

````python
# Create lexical_scores()
def lexical_scores(question, db):
    question_terms = set(preprocess_question(question))
    results = []

    for i, table in enumerate(db["table_names"]):
        for column in db["column_names"]:
            if column[0] != i:
                continue

            column_name = column[1]

            # Skip the special wildcard column "*"
            if column_name == "*":
                continue

            full_name = f"{table}.{column_name}"

            # Normalize schema column words
            schema_words = preprocess_question(column_name)

            # Include table name in matching
            schema_words += preprocess_question(table)

            schema_terms = set(schema_words)

            # Exact overlap between normalized terms
            overlap = question_terms.intersection(schema_terms)

            # Score between 0 and 1
            score = len(overlap) / max(len(schema_terms), 1)

            results.append((full_name, score))

    return results
````

````python
# Create semantic_scores()
from sentence_transformers import util

def semantic_scores(question, db):
    schema_items = []

    for i, table in enumerate(db["table_names"]):
        for column in db["column_names"]:
            if column[0] == i and column[1] != "*":
                schema_items.append(
                    f"{table}.{column[1]}"
                )

    # Encode the question and schema columns
    question_embedding = embedding_model.encode(
        question,
        convert_to_tensor=True
    )

    schema_embeddings = embedding_model.encode(
        schema_items,
        convert_to_tensor=True
    )

    # Calculate cosine similarity
    scores = util.cos_sim(
        question_embedding,
        schema_embeddings
    )[0]

    return list(zip(
        schema_items,
        scores.cpu().tolist()
    ))
````

# Create hybrid_scores()
Hybrid Score=αL+(1−α)S
Where:

L = lexical score

S = semantic score

α=0.4, giving lexical matching 40% weight and semantic matching 60%.

````python
def hybrid_scores(question, db, alpha=0.4):
    lexical_results = dict(lexical_scores(question, db))
    semantic_results = dict(semantic_scores(question, db))

    results = []

    for column in semantic_results:
        lexical = lexical_results.get(column, 0.0)
        semantic = semantic_results[column]

        hybrid = (
            alpha * lexical
            + (1 - alpha) * semantic
        )

        results.append((
            column,
            lexical,
            semantic,
            hybrid
        ))

    results.sort(
        key=lambda x: x[3],
        reverse=True
    )

    return results
````

````python
def expand_relationships(db, selected_columns):

    expanded_columns = set(selected_columns)
    relationships = []

    selected_tables = {
        column.split(".", 1)[0]
        for column in selected_columns
    }

    # Only expand relationships when more than one table
    # is already relevant to the question.
    if len(selected_tables) <= 1:
        return sorted(expanded_columns), relationships

    for fk in db["foreign_keys"]:

        # Normalized names - used for checking selected tables
        col1 = db["column_names"][fk[0]]
        col2 = db["column_names"][fk[1]]

        table1 = db["table_names"][col1[0]]
        table2 = db["table_names"][col2[0]]

        full_col1 = f"{table1}.{col1[1]}"
        full_col2 = f"{table2}.{col2[1]}"

        if table1 in selected_tables or table2 in selected_tables:

            # Add normalized columns for the retrieval pipeline
            expanded_columns.add(full_col1)
            expanded_columns.add(full_col2)

            # Original database identifiers for SQL generation
            original_col1 = db["column_names_original"][fk[0]]
            original_col2 = db["column_names_original"][fk[1]]

            original_table1 = db["table_names_original"][original_col1[0]]
            original_table2 = db["table_names_original"][original_col2[0]]

            relationship = (
                f"{original_table1}.{original_col1[1]} = "
                f"{original_table2}.{original_col2[1]}"
            )

            relationships.append(relationship)

    return sorted(expanded_columns), relationships
````

````python
# Create the reduced schema context
def create_schema_context(db, selected_columns):
    selected_set = set(selected_columns)

    schema_text = f"Database: {db['db_id']}\n\n"
    schema_text += "Relevant Tables and Columns:\n"

    for i, table in enumerate(db["table_names"]):
        table_columns = []

        for column in db["column_names"]:
            if column[0] != i or column[1] == "*":
                continue

            full_name = f"{table}.{column[1]}"

            if full_name in selected_set:
                table_columns.append(column[1])

        if table_columns:
            schema_text += f"\nTable: {table}\n"
            for column in table_columns:
                schema_text += f"- {column}\n"

    relationships = []

    for fk in db["foreign_keys"]:
        col1 = db["column_names"][fk[0]]
        col2 = db["column_names"][fk[1]]

        table1 = db["table_names"][col1[0]]
        table2 = db["table_names"][col2[0]]

        full_col1 = f"{table1}.{col1[1]}"
        full_col2 = f"{table2}.{col2[1]}"

        if full_col1 in selected_set and full_col2 in selected_set:
            relationships.append(
                f"{full_col1} = {full_col2}"
            )

    if relationships:
        schema_text += "\nRelationships:\n"
        for relationship in relationships:
            schema_text += f"- {relationship}\n"

    return schema_text
````

````python
def generate_sql(question, schema_context):

    messages = [
        {
            "role": "system",
            "content": (
                "You are an expert SQLite SQL generation assistant.\n\n"

                "Your task is to convert the user's natural language question "
                "into exactly ONE valid SQLite SQL query.\n\n"

                "STRICT RULES:\n"

                "1. Return ONLY the SQL query. Do not provide explanations, "
                "comments, markdown, or multiple queries.\n"

                "2. Use ONLY the tables and columns explicitly provided in the schema.\n"

                "3. Copy table names and column names EXACTLY as they appear "
                "in the schema.\n"

                "4. NEVER invent a table or column name.\n"

                "5. NEVER change an identifier. If a column contains an underscore, "
                "keep the underscore exactly.\n"

                "6. Do NOT replace underscores with spaces. For example, "
                "`song_name` must remain `song_name`, not `song name`.\n"

                "7. Do NOT assume that a table has a column such as `id`, `pet_id`, "
                "`pet_type`, etc. unless that exact column is present in the schema.\n"

                "8. Use the simplest SQL query that correctly answers the question.\n"

                "9. Use the minimum number of tables required to answer the question. "
                "Do NOT add a JOIN merely because a foreign-key relationship is provided.\n"

                "10. Before adding a JOIN, check whether the question actually requires "
                "information, filtering, grouping, or conditions from the other table. "
                "If the question can be answered using one table, use only that table.\n"

                "11. Foreign-key relationships are provided only to help you construct "
                "JOIN conditions when a JOIN is genuinely necessary. "
                "They do NOT imply that the related table must be included.\n"

                "12. Use SQLite-compatible SQL syntax only.\n"

                "13. Do NOT use unsupported functions such as YEAR(). "
                "Use SQLite-compatible expressions instead.\n"

                "14. For INTERSECT, UNION, or EXCEPT, write valid SQLite syntax "
                "and avoid unnecessary parentheses around individual SELECT statements.\n"

                "15. Before returning the SQL, verify that every table name and "
                "every column name exists in the provided schema.\n"

                "16. For aggregation questions, use the table that directly contains "
                "the requested attribute whenever possible. Do not introduce another "
                "table unless it is required for filtering, grouping, or obtaining "
                "information that is not available in the selected table.\n"

                "17. Do not add a JOIN simply because two tables are related. "
                "If a question asks for COUNT, MIN, MAX, AVG, SUM, or another "
                "aggregate that can be answered from one table, prefer that "
                "single-table query.\n"

                "18. Carefully preserve the logical meaning of words such as "
                "'both', 'either', 'all', 'none', 'only', and 'at least'. "
                "Do not replace a condition requiring multiple independent criteria "
                "with a logically different condition such as BETWEEN.\n"

                "19. Prefer the simplest query that directly matches the wording "
                "of the question. Avoid unnecessary JOINs, GROUP BY clauses, "
                "aliases, or subqueries.\n"

                "20. Do not select columns that are not requested by the question "
                "unless they are required for filtering, grouping, ordering, "
                "or constructing the result.\n"

                "21. When the question asks for information from only one table, "
                "do not join another table even if a foreign-key relationship exists.\n"

                "22. When a JOIN is necessary, use the provided foreign-key "
                "relationship to construct the correct JOIN condition.\n"

                "23. For counting records, use COUNT(*) when the question asks "
                "for the number of records and no specific column needs to be counted. "
                "Use COUNT(column) or COUNT(DISTINCT column) only when the question "
                "requires it.\n"

                "24. For conditions involving multiple values or categories, "
                "translate the natural-language logic exactly. Do not simplify "
                "the condition unless the simplified expression is logically equivalent.\n"

                "25. Before returning the final SQL, mentally verify:\n"
                "    - every table exists in the schema,\n"
                "    - every column exists in the schema,\n"
                "    - every JOIN is necessary,\n"
                "    - the WHERE conditions match the question,\n"
                "    - the aggregation matches the question,\n"
                "    - the SQL is valid SQLite syntax.\n"


                "26. Distinguish columns by meaning, not just similar names. "
                "For example, use Song_Name for a song's name and Name for a singer's name.\n"

                "27. Every alias referenced in SQL must be declared in FROM or JOIN.\n"


            )
        },
        {
            "role": "user",
            "content": f"""
Database schema:
{schema_context}

Question:
{question}

Generate the SQL query:
"""
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=200,
            do_sample=False
        )

    generated_sql = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    )

    return generated_sql.strip()
````

````python
# Build the main pipeline function
def run_pipeline(example, db, top_k=5):
    question = example["question"]

    # 1. Lexical + semantic + hybrid scoring
    hybrid_results = hybrid_scores(question, db)

    # 2. Select relevant columns
    selected_columns = select_columns(
        hybrid_results,
        top_k=top_k
    )

    # 3. Expand relationships
    expanded_columns,relationship = expand_relationships(
        db,
        selected_columns
    )

    # 4. Create reduced schema context
    schema_context = create_schema_context(
        db,
        expanded_columns

    )

    # 5. Generate SQL using Qwen
    predicted_sql = generate_sql(
        question,
        schema_context
    )

    # 6. Save result fields
    return {
        "db_id": example["db_id"],
        "question": question,
        "schema_context": schema_context,
        "predicted_sql": predicted_sql,
        "gold_sql": example["query"]
    }
````

````python

def create_schema_context(db, selected_columns):
    clean_tables = db["table_names"]
    original_tables = db["table_names_original"]
    clean_columns = db["column_names"]
    original_columns = db["column_names_original"]

    selected_set = set(selected_columns)
    schema_text = f"Database: {db['db_id']}\n\n"
    schema_text += "Relevant Tables and Columns:\n"

    for i, original_table in enumerate(original_tables):
        table_columns = []

        for j, original_col in enumerate(original_columns):
            table_idx, original_col_name = original_col

            if table_idx != i or original_col_name == "*":
                continue

            clean_col_name = clean_columns[j][1]
            clean_full_name = f"{clean_tables[i]}.{clean_col_name}"

            if clean_full_name in selected_set:
                table_columns.append(original_col_name)

        if table_columns:
            schema_text += f"\nTable: {original_table}\n"
            for col in table_columns:
                schema_text += f"- {col}\n"


    relationships = []

    for fk1, fk2 in db["foreign_keys"]:
        table1_idx, col1 = original_columns[fk1]
        table2_idx, col2 = original_columns[fk2]

        clean_col1 = clean_columns[fk1][1]
        clean_col2 = clean_columns[fk2][1]

        clean_full1 = f"{clean_tables[table1_idx]}.{clean_col1}"
        clean_full2 = f"{clean_tables[table2_idx]}.{clean_col2}"

        if clean_full1 in selected_set and clean_full2 in selected_set:
            relationships.append(
                f"{original_tables[table1_idx]}.{col1} = "
                f"{original_tables[table2_idx]}.{col2}"
            )

    if relationships:
        schema_text += "\nForeign-Key Relationships:\n"
        for rel in relationships:
            schema_text += f"- {rel}\n"

    return schema_text
````

````python
# PHASE 7.2: Initial clarification decision logic


def should_clarify(question):
    question = question.strip().lower()

    # Empty or incomplete input
    if not question or question.endswith("..."):
        return True, "The question appears incomplete."

    # Vague requests that need an explicit criterion
    ambiguous_phrases = [
        "best student",
        "best singer",
        "best stadium",
        "best employee",
        "best department",
        "top employees",
        "most successful"
    ]

    for phrase in ambiguous_phrases:
        if phrase in question:
            return True, (
                f"The term '{phrase}' needs a clearer criterion."
            )

    return False, "The question can proceed to SQL generation."
````

````python
# PHASE 7.3: Targeted Clarification Question Generator

def generate_clarification(question):
    question_lower = question.strip().lower()

    clarification_rules = {
        "best student": (
            "What criterion should define the best student — "
            "highest marks, attendance, or another measure?"
        ),
        "top employees": (
            "How should employees be ranked — by salary, "
            "performance, experience, or another criterion?"
        ),
        "best employee": (
            "What criterion should define the best employee?"
        ),
        "best department": (
            "How should departments be compared — by employee "
            "count, average salary, or another measure?"
        ),
        "most successful": (
            "What metric should define success for this query?"
        )
    }

    if not question_lower:
        return "What would you like to find out?"

    if question_lower.endswith("..."):
        return "Could you complete your question?"

    for phrase, clarification in clarification_rules.items():
        if phrase in question_lower:
            return clarification

    return None


# Test the clarification generator
for case in phase7_test_cases:
    if should_clarify(case["question"])[0]:
        print("\nQuestion:", case["question"])
        print("Clarification:", generate_clarification(case["question"]))
````

````python
# PHASE 7.4: Unified Clarification Pipeline

def clarification_pipeline(question):
    decision, reason = should_clarify(question)

    if decision:
        clarification = generate_clarification(question)

        return {
            "status": "clarification_required",
            "original_question": question,
            "reason": reason,
            "clarification_question": clarification
        }

    return {
        "status": "ready_for_sql",
        "original_question": question,
        "reason": reason,
        "clarification_question": None
    }


# Test the unified pipeline
test_questions = [
    "Show me the best student.",
    "Show students from...",
    "List all singers.",
    "Count the students in each department."
]

for question in test_questions:
    result = clarification_pipeline(question)

    print("\nQuestion:", result["original_question"])
    print("Status:", result["status"])
    print("Reason:", result["reason"])
    print("Clarification:", result["clarification_question"])
    print("-" * 60)
````

````python

# PHASE 7.6: Integrated clarification + SQL generation

def queryclarify_ai(
    question,
    db_id,
    clarification_answer=None,
    top_k=5
):
    # Step 1: Validate the database
    if db_id not in schema_lookup:
        return {
            "status": "error",
            "message": f"Unknown database: {db_id}",
            "sql": None
        }

    db = schema_lookup[db_id]

    # Step 2: Analyze ambiguity using the actual schema
    analysis = analyze_ambiguity(question, db_id)

    if analysis.get("status") == "analysis_error":
        return {
            "status": "analysis_error",
            "message": "The ambiguity analyzer did not return valid JSON.",
            "details": analysis.get("raw_response"),
            "sql": None
        }

    # Step 3: Validate the analyzer's decision
    decision = validate_ambiguity_result(
        analysis,
        question,
        db_id
    )

    if decision.get("status") == "error":
        return {
            "status": "error",
            "message": decision["message"],
            "sql": None
        }

    # Step 4: Ask clarification if required
    if decision["needs_clarification"]:
        if not clarification_answer or not clarification_answer.strip():
            return {
                "status": "clarification_required",
                "original_question": question,
                "clarification_question": decision.get(
                    "clarification_question"
                ),
                "reason": decision.get("reason"),
                "sql": None
            }

        # Step 5: Incorporate the user's answer
        resolved = incorporate_clarification(
            question,
            clarification_answer
        )

        question_for_sql = resolved["resolved_question"]

    else:
        question_for_sql = question

    # Step 6: Run the existing Phase 6 retrieval pipeline
    scores = hybrid_scores(question_for_sql, db)
    selected_columns = select_columns(scores, top_k=top_k)

    schema_context = create_schema_context(
        db,
        selected_columns
    )

    # Step 7: Generate SQL
    sql = clean_sql(
        generate_sql(question_for_sql, schema_context)
    )

    clarification_answered = (
    decision["needs_clarification"]
    and clarification_answer is not None
    and bool(clarification_answer.strip())
)

    return {
        "status": "sql_generated",
        "original_question": question,
        "question_for_sql": question_for_sql,
        "db_id": db_id,
        "selected_columns": selected_columns,
        "schema_context": schema_context,
        "ambiguity_analysis": decision,
        "sql": sql,
        "clarification_answered": clarification_answered
    }
````

````python

import json

def analyze_ambiguity(question, db_id):
    if db_id not in schema_lookup:
        raise ValueError(f"Unknown Spider database: {db_id}")

    db = schema_lookup[db_id]

    # Use the real Spider schema as grounding context
    schema_context = create_schema_context(
        db,
        extract_schema_columns(db)
    )

    prompt = f"""
You are the ambiguity analyzer for QueryClarify AI,
an interactive Text-to-SQL system.

Analyze the user's question using the database schema.

DATABASE SCHEMA:
{schema_context}

USER QUESTION:
{question}

RULES:
1. Do not assume an unstated ranking or comparison criterion.
2. Determine whether missing information could materially
   change the SQL or the answer.
3. Do not ask unnecessary questions when the intended
   operation and criterion are sufficiently clear.
4. Use only entities and attributes supported by the schema.
5. If multiple interpretations are plausible, ask one
   concise question about the most important missing detail.
6. Do not generate SQL.

Return ONLY a valid JSON object with these keys:
{{
  "entity": "requested entity or null",
  "operation": "requested operation or null",
  "criterion": "explicit criterion or null",
  "needs_clarification": true,
  "reason": "short explanation",
  "missing_information": "missing detail or null",
  "clarification_question": "question or null"
}}

Replace the example values with your analysis.
"""

    messages = [
        {
            "role": "system",
            "content": "You analyze ambiguity in database questions. Return JSON only."
        },
        {
            "role": "user",
            "content": prompt
        }
    ]


    inputs = tokenizer.apply_chat_template(
        messages,
        tokenize=True,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True
    )

    # Move token IDs and attention mask to the model device
    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    outputs = model.generate(
        **inputs,
        max_new_tokens=220,
        do_sample=False,
        pad_token_id=tokenizer.eos_token_id
    )

    # Extract only newly generated tokens
    input_length = inputs["input_ids"].shape[-1]
    generated = outputs[0][input_length:]

    response = tokenizer.decode(
        generated,
        skip_special_tokens=True
    ).strip()


    # Parse the model's structured response
    try:
        result = json.loads(response)
    except json.JSONDecodeError:
        return {
            "status": "analysis_error",
            "raw_response": response
        }

    result["status"] = "analysis_complete"
    return result
````

````python
def normalize_ambiguity_result(result):
    result = result.copy()

    # Standardize the field name
    if "needs_clarification" not in result:
        result["needs_clarification"] = result.get(
            "needs_clarity", False
        )

    result.pop("needs_clarity", None)

    # Convert string "null" to Python None
    for key in [
        "entity",
        "operation",
        "criterion",
        "missing_information",
        "clarification_question"
    ]:
        value = result.get(key)
        if isinstance(value, str) and value.strip().lower() == "null":
            result[key] = None

    result["needs_clarification"] = (
        result["needs_clarification"] is True
    )

    if not result["needs_clarification"]:
        result["clarification_question"] = None
        result["missing_information"] = None

    return result
````

````python

import re

def validate_ambiguity_result(result, question, db_id):
    if db_id not in schema_lookup:
        return {
            "status": "error",
            "message": f"Unknown database: {db_id}"
        }

    result = normalize_ambiguity_result(result)
    schema_columns = extract_schema_columns(schema_lookup[db_id])
    question_lower = question.lower().strip()

    # Identify explicit comparison criteria supported by the schema.
    concept_map = {
        "age": ["oldest", "youngest", "older", "younger"],
        "year": ["earliest", "latest", "newest"],
        "capacity": ["largest capacity", "highest capacity"]
    }

    supported_criteria = []

    for column in schema_columns:
        column_lower = column.lower()

        for concept, terms in concept_map.items():
            if concept in column_lower and any(
                term in question_lower for term in terms
            ):
                supported_criteria.append(column)

    # Strong textual signals that the requested operation is explicit.
    explicit_operation = bool(re.search(
        r"\b(how many|count|number of|list|show|find|give me|"
        r"return|what is|what are|which|display)\b",
        question_lower
    ))

    # Subjective requests still need a criterion when none is supplied.
    subjective_without_criterion = bool(re.search(
        r"\b(best|worst|better|worse|most suitable|most popular)\b",
        question_lower
    ))

    # Conservative override: explicit operation + no unresolved
    # subjective criterion. Do not override just because a column exists.
    if explicit_operation and not subjective_without_criterion:
        result["needs_clarification"] = False
        result["clarification_question"] = None
        result["missing_information"] = None
        result["reason"] = (
            "The question states a recognizable database operation; "
            "no subjective criterion requiring clarification was detected."
        )

    # A schema-supported criterion also resolves a comparison request.
    if supported_criteria:
        result["needs_clarification"] = False
        result["clarification_question"] = None
        result["missing_information"] = None
        result["reason"] = (
            "The schema contains a relevant comparison attribute: "
            + ", ".join(sorted(set(supported_criteria)))
        )

    result["status"] = "validated"
    result["schema_validation"] = {
        "supported_criteria": sorted(set(supported_criteria)),
        "explicit_operation_detected": explicit_operation,
        "subjective_without_criterion": subjective_without_criterion
    }

    return result
````

````python

def find_relevant_schema_columns(question, schema_columns):
    """
    Find schema columns whose names or concepts appear
    relevant to the user's question.
    """
    question_lower = question.lower()
    matches = []

    concept_map = {
        "age": ["age", "oldest", "youngest", "older", "younger"],
        "country": ["country", "nationality", "from"],
        "year": ["year", "annual", "release"],
        "name": ["name", "called", "named"],
        "capacity": ["capacity", "seats"],
    }

    for column in schema_columns:
        column_lower = column.lower()
        column_words = column_lower.replace("_", " ").replace(".", " ")

        # Direct column-name match
        if any(
            word in question_lower
            for word in column_words.split()
            if len(word) > 2
        ):
            matches.append(column)
            continue

        # Concept-based matching
        for concept, terms in concept_map.items():
            if (
                concept in column_lower
                and any(term in question_lower for term in terms)
            ):
                matches.append(column)
                break

    return list(dict.fromkeys(matches))
````

````python

import os
import json
import time
import sqlite3
import traceback

database_path = (
    "/content/drive/MyDrive/TextToSQL_Project/"
    "spider_data/spider_data/database"
)

output_path = (
    "/content/drive/MyDrive/TextToSQL_Project/"
    "phase6_phase7_eval_sample50.json"
)

# Resume safely if this evaluation file already exists.
if os.path.exists(output_path):
    with open(output_path, "r", encoding="utf-8") as f:
        combined_results = json.load(f)
    completed_indices = {r["index"] for r in combined_results}
else:
    combined_results = []
    completed_indices = set()

print("Already completed:", len(completed_indices), "/ 50")

for i, record in enumerate(sample, start=1):
    idx = record["index"]

    if idx in completed_indices:
        continue

    question = record["question"]
    db_id = record["db_id"]
    db_file = os.path.join(database_path, db_id, f"{db_id}.sqlite")

    result = {
        "index": idx,
        "db_id": db_id,
        "question": question,
        "gold_sql": record["gold_sql"],
        "baseline_sql": record["predicted_sql"],
        "baseline_execution_success": record.get("execution_success"),
        "baseline_result_match": record.get("result_match"),
    }

    try:
        pipeline_output = queryclarify_ai(question, db_id)

        result["pipeline_status"] = pipeline_output.get("status")
        result["clarification_question"] = pipeline_output.get(
            "clarification_question"
        )
        result["clarification_reason"] = pipeline_output.get("reason")
        result["predicted_sql"] = pipeline_output.get("sql")
        result["clarification_answered"] = pipeline_output.get(
            "clarification_answered", False
        )

        sql = result["predicted_sql"]

        if pipeline_output.get("status") == "clarification_required":
            result["execution_success"] = None
            result["result_match"] = None
            result["evaluation_note"] = "Clarification required; no SQL generated."

        elif not sql:
            result["execution_success"] = False
            result["result_match"] = False
            result["evaluation_note"] = "No SQL generated."

        else:
            # Execute the predicted SQL on its matching Spider database.
            try:
                uri = f"file:{db_file}?mode=ro"
                with sqlite3.connect(uri, uri=True, timeout=10) as conn:
                    predicted_rows = conn.execute(sql).fetchall()

                result["execution_success"] = True
                result["result_match"] = None

                # Execute the gold SQL for result comparison.
                try:
                    with sqlite3.connect(uri, uri=True, timeout=10) as conn:
                        gold_rows = conn.execute(
                            record["gold_sql"]
                        ).fetchall()

                    # Compare row values independent of row order.
                    # Sorting by repr supports mixed SQLite value types.
                    result["result_match"] = (
                        sorted(map(repr, predicted_rows))
                        == sorted(map(repr, gold_rows))
                    )
                except Exception as e:
                    result["gold_execution_error"] = str(e)

            except Exception as e:
                result["execution_success"] = False
                result["result_match"] = False
                result["execution_error"] = str(e)

    except Exception as e:
        result["pipeline_status"] = "pipeline_error"
        result["pipeline_error"] = str(e)
        result["execution_success"] = False
        result["result_match"] = False

    combined_results.append(result)
    completed_indices.add(idx)

    # Save after every question, so a disconnect doesn't lose all progress.
    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(combined_results, f, indent=2, ensure_ascii=False)

    print(
        f"[{len(combined_results)}/50] index={idx} | "
        f"status={result.get('pipeline_status')} | "
        f"execution={result.get('execution_success')} | "
        f"match={result.get('result_match')}"
    )

print("\nEvaluation saved to:", output_path)
print("Total completed:", len(combined_results))
````

````python
import os
import json
import sqlite3

# ==========================================
# QUERYCLARIFY AI — UPDATED 50-QUESTION EVALUATION
# ==========================================

BASE_DIR = "/content/drive/MyDrive/TextToSQL_Project"

OLD_RESULTS_FILE = os.path.join(
    BASE_DIR, "phase6_phase7_eval_sample50.json"
)

DECISIONS_FILE = os.path.join(
    BASE_DIR, "phase7_updated_decisions_sample50.json"
)

OUTPUT_FILE = os.path.join(
    BASE_DIR, "phase6_phase7_updated_eval_sample50.json"
)

DB_DIR = os.path.join(
    BASE_DIR, "spider_data", "spider_data", "database"
)

# Check that required files exist
for path in [OLD_RESULTS_FILE, DECISIONS_FILE]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Required file not found: {path}")

with open(OLD_RESULTS_FILE, "r", encoding="utf-8") as f:
    old_results = json.load(f)

with open(DECISIONS_FILE, "r", encoding="utf-8") as f:
    decisions = json.load(f)

print("Original sample:", len(old_results))
print("Updated decisions:", len(decisions))

# Handle either a list of records or a dictionary containing records
if isinstance(old_results, dict):
    old_results = old_results.get("results", old_results.get("records", []))

if isinstance(decisions, dict):
    decisions = decisions.get("decisions", decisions.get("results", []))

if not isinstance(old_results, list) or not isinstance(decisions, list):
    raise ValueError("The evaluation files do not contain lists of records.")

# Match decisions to questions using their original dataset index
decision_map = {
    str(item["index"]): item
    for item in decisions
    if "index" in item
}

# Resume previous work if this updated output already exists.
# This does not modify the frozen Phase 6 baseline.
if os.path.exists(OUTPUT_FILE):
    with open(OUTPUT_FILE, "r", encoding="utf-8") as f:
        completed_results = json.load(f)

    if isinstance(completed_results, dict):
        completed_results = completed_results.get(
            "results", completed_results.get("records", [])
        )
else:
    completed_results = []

completed_indices = {
    str(item["index"])
    for item in completed_results
    if "index" in item
}

print("Already completed:", len(completed_indices))
print("Output:", OUTPUT_FILE)


# ==========================================
# SQL EXECUTION
# ==========================================

def execute_sql_readonly(db_id, sql):
    db_path = os.path.join(DB_DIR, db_id, db_id + ".sqlite")

    if not os.path.exists(db_path):
        return False, None, f"Database file not found: {db_path}"

    try:
        # Read-only connection prevents database modifications
        uri = "file:" + db_path.replace("\\", "/") + "?mode=ro"
        conn = sqlite3.connect(uri, uri=True, timeout=10)

        try:
            cursor = conn.cursor()
            cursor.execute(sql)
            rows = cursor.fetchall()
            return True, rows, None
        finally:
            conn.close()

    except Exception as e:
        return False, None, str(e)


def rows_match(rows_a, rows_b):
    # Compare rows without depending on their returned order.
    # Duplicate rows are preserved.
    try:
        return sorted(map(repr, rows_a)) == sorted(map(repr, rows_b))
    except Exception:
        return False


# ==========================================
# RUN UPDATED PIPELINE
# ==========================================

for position, item in enumerate(old_results, start=1):

    original_index = str(item["index"])

    if original_index in completed_indices:
        continue

    decision = decision_map.get(original_index)

    if decision is None:
        print(f"Skipping index {original_index}: decision not found.")
        continue

    question = item["question"]
    db_id = item["db_id"]
    gold_sql = item["gold_sql"]

    needs_clarification = decision.get(
        "needs_clarification",
        decision.get("needs_clarification_required", False)
    )

    # Record a clarification request instead of generating SQL
    if needs_clarification:
        result = {
            "index": item["index"],
            "db_id": db_id,
            "question": question,
            "gold_sql": gold_sql,
            "status": "clarification_required",
            "clarification_question": decision.get(
                "clarification_question"
            ),
            "missing_information": decision.get(
                "missing_information"
            ),
            "baseline_execution_success": item.get(
                "baseline_execution_success",
                item.get("execution_success")
            ),
            "baseline_result_match": item.get(
                "baseline_result_match",
                item.get("result_match")
            )
        }

    else:
        try:
            # Phase 6: hybrid schema retrieval
            db_schema = schema_lookup[db_id]

            hybrid_results = hybrid_scores(
                question, db_schema, alpha=0.4
            )

            selected_columns = select_columns(
                hybrid_results, top_k=5
            )

            schema_context = create_schema_context(
                db_schema, selected_columns
            )

            predicted_sql = clean_sql(
                generate_sql(question, schema_context)
            )

            # Execute both predicted and gold SQL
            pred_success, pred_rows, pred_error = execute_sql_readonly(
                db_id, predicted_sql
            )

            gold_success, gold_rows, gold_error = execute_sql_readonly(
                db_id, gold_sql
            )

            result_match = (
                pred_success
                and gold_success
                and rows_match(pred_rows, gold_rows)
            )

            result = {
                "index": item["index"],
                "db_id": db_id,
                "question": question,
                "gold_sql": gold_sql,
                "predicted_sql": predicted_sql,
                "status": "sql_generated",
                "execution_success": pred_success,
                "execution_error": pred_error,
                "gold_execution_success": gold_success,
                "gold_execution_error": gold_error,
                "result_match": result_match,
                "clarification_question": None,
                "baseline_execution_success": item.get(
                    "baseline_execution_success",
                    item.get("execution_success")
                ),
                "baseline_result_match": item.get(
                    "baseline_result_match",
                    item.get("result_match")
                )
            }

        except Exception as e:
            result = {
                "index": item["index"],
                "db_id": db_id,
                "question": question,
                "gold_sql": gold_sql,
                "status": "pipeline_error",
                "pipeline_error": str(e)
            }

    completed_results.append(result)
    completed_indices.add(original_index)

    # Save after each question so progress is not lost
    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        json.dump(completed_results, f, indent=2, ensure_ascii=False)

    print(
        f"[{len(completed_results)}/{len(old_results)}] "
        f"Index {original_index}: {result['status']}"
    )


# ==========================================
# FINAL SUMMARY
# ==========================================

clarification_count = sum(
    r.get("status") == "clarification_required"
    for r in completed_results
)

generated = [
    r for r in completed_results
    if r.get("status") == "sql_generated"
]

execution_successes = sum(
    r.get("execution_success") is True for r in generated
)

result_matches = sum(
    r.get("result_match") is True for r in generated
)

print("\n========== UPDATED EVALUATION SUMMARY ==========")
print("Completed:", len(completed_results), "/", len(old_results))
print("Clarification requested:", clarification_count)
print("SQL generated:", len(generated))
print("Pipeline errors:", sum(
    r.get("status") == "pipeline_error"
    for r in completed_results
))
print("SQL execution success:", execution_successes, "/", len(generated))
print("Result matches:", result_matches, "/", len(generated))
print("Saved to:", OUTPUT_FILE)
````

````python

# ============================================================
# QUERYCLARIFY AI — FULL SPIDER VALIDATION EVALUATION
# Phase 6 + Phase 7 | 1,034 questions
# ============================================================

import os
import json
import sqlite3
from pathlib import Path

BASE_DIR = "/content/drive/MyDrive/TextToSQL_Project"
DATA_DIR = os.path.join(BASE_DIR, "spider_data", "spider_data")
DEV_FILE = os.path.join(DATA_DIR, "dev.json")
DB_DIR = os.path.join(DATA_DIR, "database")

# Frozen baseline: READ ONLY. Never overwrite this file.
BASELINE_FILE = os.path.join(
    BASE_DIR, "phase6_eval_1034_new.json"
)

# New full evaluation output
OUTPUT_FILE = os.path.join(
    BASE_DIR, "phase7_full_eval_1034.json"
)

# ------------------------------------------------------------
# 1. Verify prerequisites
# ------------------------------------------------------------

required_functions = [
    "analyze_ambiguity",
    "validate_ambiguity_result",
    "hybrid_scores",
    "select_columns",
    "create_schema_context",
    "generate_sql",
    "clean_sql",
    "extract_schema_columns",
]

missing_functions = [
    name for name in required_functions
    if name not in globals() or not callable(globals()[name])
]

if missing_functions:
    raise RuntimeError(
        "Run the notebook cells defining these functions first: "
        + ", ".join(missing_functions)
    )

if "schema_lookup" not in globals():
    raise RuntimeError("schema_lookup is missing. Run your schema setup cell.")

for path in [DEV_FILE, BASELINE_FILE]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Required file missing: {path}")

with open(DEV_FILE, "r", encoding="utf-8") as f:
    dev_data = json.load(f)

with open(BASELINE_FILE, "r", encoding="utf-8") as f:
    baseline_data = json.load(f)

if not isinstance(dev_data, list) or len(dev_data) != 1034:
    raise ValueError(
        f"Expected 1,034 Spider dev questions; found "
        f"{len(dev_data) if isinstance(dev_data, list) else 'non-list data'}."
    )

if isinstance(baseline_data, dict):
    baseline_data = baseline_data.get(
        "results", baseline_data.get("records", [])
    )

if not isinstance(baseline_data, list):
    raise ValueError("Unexpected baseline file structure.")

baseline_map = {
    int(record["index"]): record
    for record in baseline_data
    if "index" in record
}

print("Spider validation questions:", len(dev_data))
print("Frozen baseline records:", len(baseline_map))
print("Output file:", OUTPUT_FILE)

# ------------------------------------------------------------
# 2. Resume safely if a previous run was interrupted
# ------------------------------------------------------------

if os.path.exists(OUTPUT_FILE):
    with open(OUTPUT_FILE, "r", encoding="utf-8") as f:
        results = json.load(f)

    if isinstance(results, dict):
        results = results.get("results", results.get("records", []))

    if not isinstance(results, list):
        raise ValueError("Existing output has an unexpected format.")

    print("Resuming existing output:", len(results), "records")
else:
    results = []

completed_indices = {
    int(record["index"])
    for record in results
    if "index" in record
}

# Save an initial checkpoint before the expensive model calls.
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    json.dump(results, f, indent=2, ensure_ascii=False)

# ------------------------------------------------------------
# 3. Read-only SQLite execution
# ------------------------------------------------------------

def execute_sql_readonly(db_id, sql):
    db_path = os.path.join(DB_DIR, db_id, db_id + ".sqlite")

    if not os.path.isfile(db_path):
        return False, None, f"Database not found: {db_path}"

    conn = None

    try:
        uri = Path(db_path).as_uri() + "?mode=ro"
        conn = sqlite3.connect(uri, uri=True, timeout=15)
        cursor = conn.cursor()
        cursor.execute(sql)
        rows = cursor.fetchall()
        return True, rows, None

    except Exception as exc:
        return False, None, str(exc)

    finally:
        if conn is not None:
            conn.close()


def result_rows_match(predicted_rows, gold_rows):
    # Order-insensitive comparison, preserving duplicate rows.
    # This matches the comparison convention used in our sample tests.
    return sorted(map(repr, predicted_rows)) == sorted(map(repr, gold_rows))


# ------------------------------------------------------------
# 4. Evaluate all questions
# ------------------------------------------------------------

for index, item in enumerate(dev_data):

    if index in completed_indices:
        continue

    question = item["question"]
    db_id = item["db_id"]

    # Spider dev.json normally stores the reference SQL in "query".
    gold_sql = item.get("gold_sql", item.get("query"))

    if not isinstance(gold_sql, str) or not gold_sql.strip():
        raise ValueError(
            f"Missing reference SQL at index {index}. "
            "Inspect the dev.json record before continuing."
        )

    base = baseline_map.get(index, {})

    record = {
        "index": index,
        "db_id": db_id,
        "question": question,
        "gold_sql": gold_sql,
        "baseline_execution_success": base.get("execution_success"),
        "baseline_result_match": base.get("result_match"),
    }

    try:
        if db_id not in schema_lookup:
            raise KeyError(f"Database schema missing: {db_id}")

        # Phase 7A: analyze ambiguity and validate the decision
        ambiguity = analyze_ambiguity(question, db_id)

        if not isinstance(ambiguity, dict):
            raise TypeError(
                "analyze_ambiguity did not return a dictionary."
            )

        decision = validate_ambiguity_result(
            ambiguity, question, db_id
        )

        record["ambiguity_analysis"] = decision

        if decision.get("status") == "error":
            raise RuntimeError(
                decision.get("message", "Ambiguity validation failed.")
            )

        needs_clarification = decision.get(
            "needs_clarification", False
        )

        if needs_clarification:
            record.update({
                "status": "clarification_required",
                "clarification_question": decision.get(
                    "clarification_question"
                ),
                "missing_information": decision.get(
                    "missing_information"
                ),
                "execution_success": None,
                "result_match": None,
            })

        else:
            # Phase 6: schema retrieval and SQL generation
            db_schema = schema_lookup[db_id]

            hybrid_results = hybrid_scores(
                question, db_schema, alpha=0.4
            )

            selected_columns = select_columns(
                hybrid_results, top_k=5
            )

            schema_context = create_schema_context(
                db_schema, selected_columns
            )

            predicted_sql = clean_sql(
                generate_sql(question, schema_context)
            )

            record["predicted_sql"] = predicted_sql

            # Execute prediction and reference independently
            pred_ok, pred_rows, pred_error = execute_sql_readonly(
                db_id, predicted_sql
            )

            gold_ok, gold_rows, gold_error = execute_sql_readonly(
                db_id, gold_sql
            )

            matches = (
                pred_ok
                and gold_ok
                and result_rows_match(pred_rows, gold_rows)
            )

            record.update({
                "status": "sql_generated",
                "execution_success": pred_ok,
                "execution_error": pred_error,
                "gold_execution_success": gold_ok,
                "gold_execution_error": gold_error,
                "result_match": matches,
                "clarification_question": None,
            })

    except Exception as exc:
        record.update({
            "status": "pipeline_error",
            "pipeline_error": str(exc),
            "execution_success": None,
            "result_match": None,
        })

    results.append(record)
    completed_indices.add(index)

    # Checkpoint after every question so interrupted runs can resume.
    with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
        json.dump(results, f, indent=2, ensure_ascii=False)

    if len(results) % 10 == 0 or len(results) == len(dev_data):
        print(
            f"Progress: {len(completed_indices)}/{len(dev_data)} "
            f"| Latest index: {index} "
            f"| Status: {record['status']}"
        )

# ------------------------------------------------------------
# 5. Final metrics
# ------------------------------------------------------------

generated = [
    r for r in results if r.get("status") == "sql_generated"
]

clarifications = [
    r for r in results
    if r.get("status") == "clarification_required"
]

errors = [
    r for r in results if r.get("status") == "pipeline_error"
]

exec_successes = sum(
    r.get("execution_success") is True for r in generated
)

result_matches = sum(
    r.get("result_match") is True for r in generated
)

def percentage(numerator, denominator):
    return 100 * numerator / denominator if denominator else 0.0

print("\n========== PHASE 7 FULL EVALUATION ==========")
print("Questions completed:", len(results), "/", len(dev_data))
print("SQL generated:", len(generated))
print("Clarification requested:", len(clarifications))
print("Pipeline errors:", len(errors))

print(
    "SQL execution success:",
    f"{exec_successes}/{len(generated)}",
    f"({percentage(exec_successes, len(generated)):.2f}%)"
)

print(
    "Result matches:",
    f"{result_matches}/{len(generated)}",
    f"({percentage(result_matches, len(generated)):.2f}%)"
)

print(
    "Result matches across all 1,034 questions:",
    f"{result_matches}/{len(dev_data)}",
    f"({percentage(result_matches, len(dev_data)):.2f}%)"
)

# Paired comparison on the same questions where SQL was generated.
paired = [
    r for r in generated
    if r["index"] in baseline_map
    and isinstance(
        baseline_map[r["index"]].get("execution_success"), bool
    )
    and isinstance(
        baseline_map[r["index"]].get("result_match"), bool
    )
]

print("\n========== PAIRED BASELINE COMPARISON ==========")
print("Comparable generated-query records:", len(paired))

for field, label in [
    ("execution_success", "Updated execution success"),
    ("result_match", "Updated result matches"),
]:
    successes = sum(r.get(field) is True for r in paired)
    print(
        f"{label}: {successes}/{len(paired)} "
        f"({percentage(successes, len(paired)):.2f}%)"
    )

for field, label in [
    ("execution_success", "Baseline execution success"),
    ("result_match", "Baseline result matches"),
]:
    successes = sum(
        baseline_map[r["index"]].get(field) is True
        for r in paired
    )
    print(
        f"{label} on the same subset: {successes}/{len(paired)} "
        f"({percentage(successes, len(paired)):.2f}%)"
    )

print("\nSaved evaluation:", OUTPUT_FILE)
print("Frozen baseline was not modified.")
````

````python

# ============================================================
# QUERYCLARIFY AI — FULL SPIDER EVALUATION (1,034 QUESTIONS)
# Phase 6 + Phase 7 | Progress tracking + ETA + checkpoints
# ============================================================

import os
import json
import sqlite3
import time
from pathlib import Path
from datetime import timedelta

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/TextToSQL_Project"
DATA_DIR = os.path.join(BASE_DIR, "spider_data", "spider_data")

DEV_FILE = os.path.join(DATA_DIR, "dev.json")
DB_DIR = os.path.join(DATA_DIR, "database")

# Read-only baseline. Never overwrite this file.
BASELINE_FILE = os.path.join(
    BASE_DIR, "phase6_eval_1034_new.json"
)

# New full evaluation output
OUTPUT_FILE = os.path.join(
    BASE_DIR, "phase7_full_eval_1034.json"
)

# ------------------------------------------------------------
# 2. VERIFY PREREQUISITES
# ------------------------------------------------------------

required_functions = [
    "analyze_ambiguity",
    "validate_ambiguity_result",
    "hybrid_scores",
    "select_columns",
    "create_schema_context",
    "generate_sql",
    "clean_sql",
    "extract_schema_columns",
]

missing_functions = [
    name for name in required_functions
    if name not in globals() or not callable(globals()[name])
]

if missing_functions:
    raise RuntimeError(
        "Run the notebook cells defining these functions first: "
        + ", ".join(missing_functions)
    )

if "schema_lookup" not in globals():
    raise RuntimeError(
        "schema_lookup is missing. Run your schema setup cell first."
    )

for path in [DEV_FILE, BASELINE_FILE]:
    if not os.path.exists(path):
        raise FileNotFoundError(f"Required file not found: {path}")

with open(DEV_FILE, "r", encoding="utf-8") as f:
    dev_data = json.load(f)

with open(BASELINE_FILE, "r", encoding="utf-8") as f:
    baseline_data = json.load(f)

if not isinstance(dev_data, list) or len(dev_data) != 1034:
    raise ValueError(
        f"Expected 1,034 validation records; found "
        f"{len(dev_data) if isinstance(dev_data, list) else 'non-list data'}."
    )

if isinstance(baseline_data, dict):
    baseline_data = baseline_data.get(
        "results", baseline_data.get("records", [])
    )

if not isinstance(baseline_data, list):
    raise ValueError("Unexpected baseline file structure.")

baseline_map = {
    int(r["index"]): r
    for r in baseline_data
    if "index" in r
}

print("Spider validation questions:", len(dev_data))
print("Frozen baseline records:", len(baseline_map))
print("Output file:", OUTPUT_FILE)

# ------------------------------------------------------------
# 3. LOAD PREVIOUS PROGRESS
# ------------------------------------------------------------

if os.path.exists(OUTPUT_FILE):
    with open(OUTPUT_FILE, "r", encoding="utf-8") as f:
        results = json.load(f)

    if isinstance(results, dict):
        results = results.get(
            "results", results.get("records", [])
        )

    if not isinstance(results, list):
        raise ValueError("Existing output has an unexpected format.")

    print("Existing saved records:", len(results))
else:
    results = []

# Keep only one record per index, preserving the latest one.
results_by_index = {}

for record in results:
    if "index" in record:
        results_by_index[int(record["index"])] = record

results = list(results_by_index.values())

completed_indices = set(results_by_index.keys())

# Check for invalid indexes before resuming.
invalid_indices = [
    i for i in completed_indices
    if i < 0 or i >= len(dev_data)
]

if invalid_indices:
    raise ValueError(
        f"Output contains indexes outside the validation dataset: "
        f"{invalid_indices[:10]}"
    )

# ------------------------------------------------------------
# 4. LIVE PROGRESS TRACKER
# ------------------------------------------------------------

RUN_START_TIME = time.time()
QUESTIONS_AT_START = len(completed_indices)
last_progress_time = RUN_START_TIME

def format_duration(seconds):
    return str(timedelta(seconds=max(0, int(seconds))))

def show_progress(latest_index=None, status="Running"):
    global last_progress_time

    total = len(dev_data)
    completed = len(completed_indices)
    remaining = total - completed

    elapsed = time.time() - RUN_START_TIME
    newly_completed = completed - QUESTIONS_AT_START

    if newly_completed > 0:
        avg_per_question = elapsed / newly_completed
        eta_seconds = avg_per_question * remaining
        eta = format_duration(eta_seconds)
    else:
        eta = "Calculating..."

    percentage = 100 * completed / total if total else 0

    bar_width = 30
    filled = int(bar_width * completed / total) if total else 0
    bar = "█" * filled + "░" * (bar_width - filled)

    print(
        f"\n[{bar}] {percentage:.2f}%"
        f"\nCompleted : {completed}/{total}"
        f"\nRemaining : {remaining} questions"
        f"\nElapsed   : {format_duration(elapsed)}"
        f"\nETA       : {eta}"
        f"\nLatest    : index {latest_index} | {status}"
        f"\n{'-' * 58}",
        flush=True
    )

show_progress(
    status="Resuming saved progress" if completed_indices else "Starting"
)

# ------------------------------------------------------------
# 5. READ-ONLY SQLITE EXECUTION
# ------------------------------------------------------------

def execute_sql_readonly(db_id, sql):
    db_path = os.path.join(DB_DIR, db_id, db_id + ".sqlite")

    if not os.path.isfile(db_path):
        return False, None, f"Database not found: {db_path}"

    conn = None

    try:
        uri = Path(db_path).as_uri() + "?mode=ro"
        conn = sqlite3.connect(uri, uri=True, timeout=15)

        cursor = conn.cursor()
        cursor.execute(sql)
        rows = cursor.fetchall()

        return True, rows, None

    except Exception as exc:
        return False, None, str(exc)

    finally:
        if conn is not None:
            conn.close()


def result_rows_match(predicted_rows, gold_rows):
    # Order-insensitive comparison; duplicates are preserved.
    return sorted(map(repr, predicted_rows)) == sorted(map(repr, gold_rows))


# ------------------------------------------------------------
# 6. CHECKPOINT SAVING
# ------------------------------------------------------------

def save_results():
    # Write to a temporary file first, then replace the output.
    # This helps reduce the risk of a partially written JSON file.
    temp_file = OUTPUT_FILE + ".tmp"

    with open(temp_file, "w", encoding="utf-8") as f:
        json.dump(results, f, indent=2, ensure_ascii=False)

    os.replace(temp_file, OUTPUT_FILE)


# Save the existing checkpoint in normalized form.
save_results()

# ------------------------------------------------------------
# 7. RUN THE FULL EVALUATION
# ------------------------------------------------------------

for index, item in enumerate(dev_data):

    if index in completed_indices:
        continue

    question = item["question"]
    db_id = item["db_id"]
    gold_sql = item.get("gold_sql", item.get("query"))

    record = {
        "index": index,
        "db_id": db_id,
        "question": question,
        "gold_sql": gold_sql,
        "baseline_execution_success": baseline_map.get(
            index, {}
        ).get("execution_success"),
        "baseline_result_match": baseline_map.get(
            index, {}
        ).get("result_match"),
    }

    try:
        if not isinstance(gold_sql, str) or not gold_sql.strip():
            raise ValueError("Reference SQL is missing or empty.")

        if db_id not in schema_lookup:
            raise KeyError(f"Schema missing for database: {db_id}")

        # PHASE 7: Ambiguity analysis and validation
        ambiguity = analyze_ambiguity(question, db_id)

        if not isinstance(ambiguity, dict):
            raise TypeError(
                "analyze_ambiguity did not return a dictionary."
            )

        decision = validate_ambiguity_result(
            ambiguity, question, db_id
        )

        record["ambiguity_analysis"] = decision

        if decision.get("status") == "error":
            raise RuntimeError(
                decision.get("message", "Ambiguity validation failed.")
            )

        needs_clarification = decision.get(
            "needs_clarification", False
        )

        if needs_clarification:
            record.update({
                "status": "clarification_required",
                "clarification_question": decision.get(
                    "clarification_question"
                ),
                "missing_information": decision.get(
                    "missing_information"
                ),
                "execution_success": None,
                "result_match": None,
            })

        else:
            # PHASE 6: Schema retrieval and SQL generation
            db_schema = schema_lookup[db_id]

            hybrid_results = hybrid_scores(
                question, db_schema, alpha=0.4
            )

            selected_columns = select_columns(
                hybrid_results, top_k=5
            )

            schema_context = create_schema_context(
                db_schema, selected_columns
            )

            predicted_sql = clean_sql(
                generate_sql(question, schema_context)
            )

            record["predicted_sql"] = predicted_sql

            # Execute predicted SQL and reference SQL independently
            pred_ok, pred_rows, pred_error = execute_sql_readonly(
                db_id, predicted_sql
            )

            gold_ok, gold_rows, gold_error = execute_sql_readonly(
                db_id, gold_sql
            )

            matches = (
                pred_ok
                and gold_ok
                and result_rows_match(pred_rows, gold_rows)
            )

            record.update({
                "status": "sql_generated",
                "execution_success": pred_ok,
                "execution_error": pred_error,
                "gold_execution_success": gold_ok,
                "gold_execution_error": gold_error,
                "result_match": matches,
                "clarification_question": None,
            })

    except Exception as exc:
        record.update({
            "status": "pipeline_error",
            "pipeline_error": str(exc),
            "execution_success": None,
            "result_match": None,
        })

    # Save this record and mark its index completed.
    results_by_index[index] = record
    results = list(results_by_index.values())
    completed_indices.add(index)

    # Checkpoint after every question.
    save_results()

    # Update the progress display every 5 newly completed questions,
    # plus the last question. This is based on completed records,
    # not the original dataset index.
    newly_completed = len(completed_indices) - QUESTIONS_AT_START

    if (
        newly_completed % 5 == 0
        or len(completed_indices) == len(dev_data)
    ):
        show_progress(
            latest_index=index,
            status=record["status"]
        )

# ------------------------------------------------------------
# 8. FINAL METRICS
# ------------------------------------------------------------

generated = [
    r for r in results
    if r.get("status") == "sql_generated"
]

clarifications = [
    r for r in results
    if r.get("status") == "clarification_required"
]

errors = [
    r for r in results
    if r.get("status") == "pipeline_error"
]

exec_successes = sum(
    r.get("execution_success") is True for r in generated
)

result_matches = sum(
    r.get("result_match") is True for r in generated
)

def percentage(numerator, denominator):
    return 100 * numerator / denominator if denominator else 0.0

print("\n========== PHASE 7 FULL EVALUATION ==========")
print("Questions completed:", len(completed_indices), "/", len(dev_data))
print("SQL generated:", len(generated))
print("Clarification requested:", len(clarifications))
print("Pipeline errors:", len(errors))

print(
    "SQL execution success:",
    f"{exec_successes}/{len(generated)}",
    f"({percentage(exec_successes, len(generated)):.2f}%)"
)

print(
    "Result matches among generated SQL:",
    f"{result_matches}/{len(generated)}",
    f"({percentage(result_matches, len(generated)):.2f}%)"
)

print(
    "Result matches across all 1,034 questions:",
    f"{result_matches}/{len(dev_data)}",
    f"({percentage(result_matches, len(dev_data)):.2f}%)"
)

# Paired comparison: same generated questions with available baseline metrics
paired = [
    r for r in generated
    if r["index"] in baseline_map
    and isinstance(
        baseline_map[r["index"]].get("execution_success"), bool
    )
    and isinstance(
        baseline_map[r["index"]].get("result_match"), bool
    )
]

print("\n========== PAIRED BASELINE COMPARISON ==========")
print("Comparable records:", len(paired))

for field, label in [
    ("execution_success", "Updated execution success"),
    ("result_match", "Updated result matches"),
]:
    successes = sum(r.get(field) is True for r in paired)
    print(
        f"{label}: {successes}/{len(paired)} "
        f"({percentage(successes, len(paired)):.2f}%)"
    )

for field, label in [
    ("execution_success", "Baseline execution success"),
    ("result_match", "Baseline result matches"),
]:
    successes = sum(
        baseline_map[r["index"]].get(field) is True
        for r in paired
    )
    print(
        f"{label} on same subset: {successes}/{len(paired)} "
        f"({percentage(successes, len(paired)):.2f}%)"
    )

print("\nTotal elapsed this run:",
      format_duration(time.time() - RUN_START_TIME))
print("Saved evaluation:", OUTPUT_FILE)
print("Frozen Phase 6 baseline was not modified.")
````
